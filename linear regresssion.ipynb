#What is Linear regresssion
Linear regression is a statistical and machine learning method used to model the relationship between a dependent output variable and one or more independent input variables by fitting a straight line.
Core ConceptsDependent Variable (\[Y\]): The target number you want to predict or explain.
Independent Variable (\[X\]): The predictor or input feature used to make the prediction.
Line of Best Fit: A straight line drawn through data points that minimizes prediction error.
Residuals: The vertical distances between the actual data points and the fitted line


For simple linear regression (with one independent variable), the best-fit line is represented by the equation
y=mx+b

Linear Regression is a fundamental supervised learning algorithm used to model the relationship between a dependent variable and one or more independent variables. It predicts continuous values by fitting a straight line that best represents the data.

It assumes that there is a linear relationship between the input and output
Uses a best‑fit line to make predictions
Commonly used in forecasting, trend analysis, and predictive modelling

```python
# IMPORT THE NECESSARY LIBRARIES
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#GENERATE RANDOM DATA SET 
X=np.random.rand(50,1)*100
Y=3.5*X +np.random.randn(50,1)*20
```

```python
#Creating And Training The Model
model=LinearRegression()
model.fit(X,Y)
```

```python
#PREDICTING THE VALUES
Y_pred=model.predict(X)
#Using the trained model to predict output values for the input data.
```

```python
#visualizing the results 
#Plot the original data points along with the best-fit regression line.
plt.figure(figsize=(8,6))
plt.scatter(X, Y, color='blue', label='Data points')
plt.plot(X, Y_pred, color='red', label='Regression line')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Linear Regression')
plt.legend()
plt.show()
plt.grid(True)
```

```python
#Slope the Intercept 
print("slope(coefficient):",model.coef_[0][0])
print("intercept:",model.intercept_[0])


#Print the learned slope and intercept values of the regression mode
```
